Task 3: Data Quality Checks
3.1: Ensure data consistency across state and district levels

1. For each state, calculate the total number of transactions, total transaction amount, and total registered users by summing up the values from the district-level data.

2. Compare the results with the corresponding values at the state level to ensure they match.

3. Display any discrepancies found between the district-level and state-level data.

```python
District_Summary = District_Txn_Users_df.groupby(
    ["State", "Year", "Quarter"]
).agg(
    District_Transactions=("Transactions", "sum"),
    District_Amount=("Amount (INR)", "sum"),
    District_Registered_Users=("Registered Users", "sum")
).reset_index()

District_Summary
```

```python
State_Summary = State_Txn_Users_df[
    ["State", "Year", "Quarter",
     "Transactions", "Amount (INR)", "Registered Users"]
].copy()
```

```python
State_Summary.rename(
    columns={
        "Transactions": "State_Transactions",
        "Amount (INR)": "State_Amount",
        "Registered Users": "State_Registered_Users"
    },
    inplace=True
)
```

```python
Comparison = District_Summary.merge(
    State_Summary,
    on=["State", "Year", "Quarter"],
    how="outer"
)
```

```python
Comparison["Transaction_Difference"] = (
    Comparison["District_Transactions"]
    - Comparison["State_Transactions"]
)

Comparison["Amount_Difference"] = (
    Comparison["District_Amount"]
    - Comparison["State_Amount"]
)

Comparison["Users_Difference"] = (
    Comparison["District_Registered_Users"]
    - Comparison["State_Registered_Users"]
)
```

```python
Comparison["Transaction_Difference"]
```

```python
Comparison["Amount_Difference"]
```

```python
Comparison["Users_Difference"]
```

```python
Discrepancies = Comparison[
    (Comparison["Transaction_Difference"] != 0) |
    (Comparison["Amount_Difference"].abs() > 0.01) |
    (Comparison["Users_Difference"] != 0)
]

Discrepancies
```

```python
# 1. State-wise population
State_Population_df = (
    District_Demographics_df
    .groupby("State")["Population"]
    .sum()
    .reset_index()
)

# 2. Get latest registered users for each state
Latest_State_Users_df = (
    State_Txn_Users_df
    .sort_values(["Year", "Quarter"])
    .drop_duplicates("State", keep="last")
)

# 3. Select required columns
Latest_State_Users_df = Latest_State_Users_df[
    ["State", "Registered Users"]
]

# 4. Merge
State_User_Population_df = Latest_State_Users_df.merge(
    State_Population_df,
    on="State",
    how="inner"
)

# 5. Calculate ratio
State_User_Population_df["User_Population_Ratio"] = (
    State_User_Population_df["Registered Users"]
    / State_User_Population_df["Population"]
) * 100

# 6. Display
State_User_Population_df[
    ["State", "Registered Users", "Population", "User_Population_Ratio"]
].sort_values(
    "User_Population_Ratio",
    ascending=False
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14, 6))

plt.bar(
    State_User_Population_df["State"],
    State_User_Population_df["User_Population_Ratio"]
)

plt.xlabel("State")
plt.ylabel("Registered Users / Population (%)")
plt.title("Registered Users to Population Ratio by State")

plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```
